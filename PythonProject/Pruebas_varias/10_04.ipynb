```python
# Importar las librerías necesarias
import seaborn as sns  # Para gráficos estadísticos
import matplotlib.pyplot as plt  # Para personalización
import pandas as pd  # Para manejar datos
 
# Cargar dataset de ejemplo incluido en Seaborn
datos = sns.load_dataset('tips')  # Datos de propinas en restaurantes
 
print(datos.head())

# Crear gráfico de dispersión avanzado
plt.figure(figsize=(8, 6))
scatter = sns.scatterplot(
    data=datos,  # DataFrame completo
    x='total_bill',  # Variable en eje X (cuenta total)
    y='tip',  # Variable en eje Y (propina)
    hue='size',  # Color por tamaño de grupo
    size='size',  # Tamaño de puntos por grupo
    sizes=(30, 200),  # Rango de tamaños
    alpha=0.7,  # Transparencia para solapamientos
    palette='viridis'  # Paleta de colores
)
 
# Línea de referencia (propina del 15%)
plt.axline((0, 0), slope=0.15, color='red', linestyle='--', label='15%')
 
# Personalización
plt.title('Relación entre Total de la Cuenta y Propina', fontsize=14)
plt.xlabel('Total de la Cuenta ($)', fontsize=12)
plt.ylabel('Propina ($)', fontsize=12)
plt.legend(title='Tamaño del Grupo', bbox_to_anchor=(1.05, 1))  # Leyenda fuera
plt.grid(True, linestyle=':', alpha=0.5)  # Grid punteado
 
plt.tight_layout()  # Ajustar espacios
plt.show()
```

```python
propinas_promedio = datos.groupby("day", as_index=False)["tip"].mean()
conteo_horario = datos['time'].value_counts().reset_index()
conteo_horario.columns = ['time', 'count']

# Configurar figura con 3 subplots
plt.figure(figsize=(15, 5))
 
# Gráfico de líneas (izquierda)
plt.subplot(1, 3, 1)
sns.lineplot(data=propinas_promedio, x='day', y='tip', marker='o', color='green')
plt.title('Propina Promedio por Día')
plt.xticks(rotation=45)  # Rotar etiquetas 45 grados
 
# Gráfico de barras (centro)
plt.subplot(1, 3, 2)
sns.barplot(data=conteo_horario, x='time', y='count', palette=['skyblue', 'salmon'])
plt.title('Comensales por Horario')
 
# Gráfico de dispersión (derecha)
plt.subplot(1, 3, 3)
sns.scatterplot(data=datos, x='total_bill', y='tip', hue='size', palette='viridis', alpha=0.7)
plt.title('Total vs Propina')
plt.legend([], [], frameon=False)  # Ocultar leyenda
 
plt.tight_layout()  # Ajustar automáticamente
plt.show()
```
