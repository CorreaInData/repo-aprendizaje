```python
import pandas as pd # Importa la libreria Pandas
import matplotlib.pyplot as plt  # Importa la libreria matplotlib
from sklearn.preprocessing import StandardScaler # Importa el módulo StandardScaler de la librería de scikit-learn

# Cargar datos 
df = pd.read_csv('car_data.csv') # Carga los datos desde un archivo CSV y lo guarda en la variable df
variable = df['Year'] # Guarda los datos correspondientes al campo 'Year' en la variable 'variable'
# Estandarizar la variable 
scaler = StandardScaler() # Crea la instancia del método StandardScaler
variable_scaled = scaler.fit_transform(variable.values.reshape(-1, 1)) # Realiza la normalización: Primero extraelos valores de 'variable' y luego los convierte en un arreglo de formato columna, para que el método pueda funcionar. Y por último calcula la media y la deviación y realiza la estandarización con estos dos datos.
# Graficar el histograma 
plt.figure(figsize=(12, 6)) # Establece las medidas del grafico
plt.subplot(1, 2, 1) # Genera una partición en la figura y selecciona la primera parte
plt.hist(variable, bins=30, alpha=0.7, label='Original') # crea un grafico de barras a partir de 'variable"
plt.title('Histograma Original') # Asigna un titulo al gráfico
plt.subplot(1, 2, 2) # Selecciona la segunda parte de la figura
plt.hist(variable_scaled, bins=30, alpha=0.7, label='Normalizado') # crea un grafico de barras a partir de 'variable"
plt.title('Histograma Normalizado') # Asigna un titulo al gráfico
plt.show() # Muestra el gráfico
```

```python
import pandas as pd 
import matplotlib.pyplot as plt 
from sklearn.preprocessing import StandardScaler 
# Cargar datos 
df = pd.read_csv('car_data.csv') 
variables = df[['Year', 'Selling_Price']] # Reemplazar con los nombres de las variables 
# Escalar las variables 
scaler = StandardScaler() 
variables_scaled = scaler.fit_transform(variables) 
# Graficar el scatter plot 
plt.figure(figsize=(12, 6)) 
plt.subplot(1, 2, 1) 
plt.scatter(variables['Year'], variables['Selling_Price'], alpha=0.7, label='Original') 
plt.title('Datos Originales') 
plt.subplot(1, 2, 2) 
plt.scatter(variables_scaled[:, 0], variables_scaled[:, 1], alpha=0.7, label='Escalado') 
plt.title('Datos Escalados') 
plt.show()
```

```python
import pandas as pd 
import matplotlib.pyplot as plt 
from sklearn.model_selection import train_test_split 
# Cargar datos 
df = pd.read_csv('car_data.csv') 
# Crear una variable categórica para estratificar 
df['estrato'] = pd.cut(df['Fuel_Type'], bins=3, labels=['Bajo', 'Medio', 'Alto']) 
# Realizar muestreo estratificado 
X_train, X_test = train_test_split(df, test_size=0.2, stratify=df['estrato']) 
# Graficar el muestreo estratificado 
plt.figure(figsize=(8, 6)) 
df['estrato'].value_counts().plot(kind='bar', alpha=0.7) 
plt.title('Distribución de la Variable Estratificada') 
plt.xlabel('Estrato') 
plt.ylabel('Frecuencia') 
plt.show() 
```
