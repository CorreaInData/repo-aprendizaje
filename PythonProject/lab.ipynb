```python
# Importación de librerías
import matplotlib.pyplot as plt
import numpy as np
 
# Crear datos
x = np.linspace(0, 10, 100)  # 100 puntos entre 0 y 10
y = np.sin(x)  # Función seno
 
# Crear el gráfico de líneas
plt.plot(x, y, label='y = sin(x)', color='blue', linestyle='-')
plt.title("Gráfico de Líneas")
plt.xlabel("Eje X")
plt.ylabel("Eje Y")
plt.legend()
plt.grid(True)
plt.show()
 
```

```python
# Crear datos
categorias = ['A', 'B', 'C', 'D']
valores = [23, 45, 56, 78]
 
# Crear el gráfico de barras
plt.bar(categorias, valores, color='red')
plt.title("Gráfico de Barras")
plt.xlabel("Categorías")
plt.ylabel("Valores")
plt.show()
```

```python
# Crear datos
x = np.random.rand(50)  # 50 valores aleatorios
y = np.random.rand(50)  # 50 valores aleatorios
 
# Crear el gráfico de dispersión
plt.scatter(x, y, color='red', marker='o')
plt.title("Gráfico de Dispersión")
plt.xlabel("Eje X")
plt.ylabel("Eje Y")
plt.grid(True)
plt.show()
```

```python
# Crear datos
x = np.linspace(-10, 10, 100)
y = np.pow(x, 2)

plt.plot(x, y, label='y = x^2', color='brown', linestyle='-')
plt.title("Gráfico de lineas")
plt.xlabel("Eje x")
plt.ylabel("Eje y")
plt.legend()
plt.grid(True)
plt.show()
```
